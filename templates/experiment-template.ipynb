## How to use this template (delete this cell when you're done reading)

Each experiment is three cells: **setup → code → results**.
1. Write what you're testing and what you *think* will happen, before running anything.
2. Write the code and run it (**Shift+Enter**).
3. Compare what happened to your prediction and write down why.

**New experiment:** click the setup cell, Shift+click the results cell, press **C**, click the last cell, press **V**.
