## EXAMPLE: Joining strings with `+` (delete these three cells when you're done)

```python
myVar1 = 'Learning'          # a string stored in a variable
myVar2 = 'Python'
result = myVar1 + myVar2     # + on two strings joins them end to end
print(result)
```

**What it does:** Stores two words in variables, joins them, and prints the result.


**Why it works:** When `+` is used on two strings, it doesn't do math. It *concatenates*, gluing the second string onto the end of the first, exactly as they are. Python never adds a space for you.


**Output:** `LearningPython`


**Notes to self:** No space between the words. To get one, I'd need to add a `' '` string in the middle. Also, the course uses camelCase (`myVar1`), but normal Python style is snake_case (`my_var1`).
