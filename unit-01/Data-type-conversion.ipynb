## Simple Data Conversion

```python
myVar1 = 111
myVar2 = 222
result = str(myVar1) + str(myVar2)
print(result)
```

**What it does:** *if a variable is already set as an integer you can convert them into a string using the str function*


**Why it works:** *it works by converting something set as an integer into a string. That way you can use it later on.*


**Output:** *111222*


**Notes to self:** *we can also convert a string into an integer*

```python
myVar1 = 111
myVar2 = 222
result = int(myVar1) + int(myVar2)
print(result)
```
