## Example 1: Joining two strings with `+`

**What it does:** Stores two words in variables, joins them with `+`, and prints the result.

**Why:** When `+` is used on two strings, it doesn't add numbers. It *concatenates*, gluing the second string onto the end of the first. Python joins exactly what it's given, so no space gets added.

**Output:** `LearningPython` (no space)

**Note to self:** Python convention is `snake_case` (`my_var1`), not camelCase like JavaScript. The course uses `myVar1` here, which works but isn't standard Python style.

```python
myVar1 = 'Learning'          # store the string 'Learning' in a variable
myVar2 = 'Python'            # store the string 'Python' in a second variable
result = myVar1 + myVar2     # + on strings = concatenation (join end to end)
print(result)                # display the joined string: LearningPython
```
