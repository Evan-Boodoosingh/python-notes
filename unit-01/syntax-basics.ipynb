## Overview

**Big idea:** Python's core building blocks: comments, storing values in variables, working with strings, ints, and floats, converting between types, getting input, and making decisions with `if`/`elif`/`else`. Indentation is part of the syntax, not just style.

**Key terms:**
- **variable:** a name that stores a value (`grade = 85`)
- **string (`str`):** text in quotes, `'hi'` or `"hi"`
- **integer (`int`):** a whole number, `85`
- **float:** a number with a decimal, `85.5`
- **boolean (`bool`):** `True` or `False` (capitalized in Python)
- **concatenation:** joining strings end to end with `+`
- **type conversion:** changing a value's type with `int()`, `str()`, `float()`
- **condition:** an expression that's `True` or `False`, like `grade > 90`
- **block:** the indented lines that belong to a line ending in `:`
- **operator:** a symbol or word that does something with values (`+`, `==`, `and`)
- **floor division (`//`):** divide and drop the decimal
- **modulo (`%`):** the remainder after dividing
- **escape sequence:** a backslash code inside a string, like `\n` (new line) or `\t` (tab)
- **f-string:** a string starting with `f` that drops values into `{ }`
- **comment:** text after `#` that Python ignores
- **syntax / runtime / logic error:** breaks the rules / fails while running / runs but gives the wrong answer
- **exception:** an error Python raises while running, like `ValueError`

**General notes:**
- Python uses indentation (4 spaces) where JavaScript uses `{ }`. Wrong indentation changes what the code means.
- No `let` or `const`. You just assign: `name = 'Evan'`.
- Style is `snake_case` (`my_var`), not camelCase. The course uses `myVar1`, which works but isn't standard.

# Part 1: Basics

## Comments

````python
# A full-line comment: Python ignores everything after the #
print("Hello")   # a comment can also go at the end of a line
# print("This line is turned off, so it never runs")
````

**What it does:** Prints `Hello`. The comment lines are skipped.


**Why it works:** Anything after `#` on a line is ignored by Python. Use comments to explain *why* code does something, or to temporarily switch a line off ("commenting it out").


**Output:** `Hello`


**Notes to self:** JS uses `//`. In VS Code, **Cmd+/** toggles a comment on the current or selected lines.

## Variables

````python
grade = 85              # an int
name = 'Evan'           # a str (single or double quotes both work)
passed = True           # a bool: capital T, unlike JavaScript's true
print(name, grade, passed)   # print accepts several values, separated by commas
````

**What it does:** Creates three variables of different types and prints them on one line.


**Why it works:** Assigning with `=` creates the variable. Python figures out the type from the value, so there's no type keyword. When `print()` gets several values separated by commas, it puts a space between each one.


**Output:** `Evan 85 True`


**Notes to self:** Commas in `print()` add spaces automatically. `+` does not (see next section).

## Variable naming rules

````python
student_name = 'Evan'    # snake_case: the Python standard
age2 = 30                 # numbers are fine, just not first
_count = 5                # can start with an underscore
Score = 90                # legal, but Score and score are DIFFERENT variables
score = 80
print(Score, score)
````

**What it does:** Creates variables with valid names and shows that capitalization matters.


**Why it works:** Names can use letters, numbers, and underscores, but can't start with a number, can't contain spaces or dashes, and can't be a Python keyword like `if`, `else`, or `True`. (`print` is technically allowed as a name, but then the print function stops working.) Names are case-sensitive.


**Output:** `90 80`


**Notes to self:** These crash with SyntaxError: `2age = 30`, `my name = 'x'`, `my-name = 'x'`. Pick names that say what the value is: `grade`, not `g`.

# Part 2: Strings

## Joining strings with `+`

````python
myVar1 = 'Learning'
myVar2 = 'Python'
print(myVar1 + myVar2)          # + glues strings together exactly as they are
print(myVar1 + ' ' + myVar2)    # ' ' is a string containing one space
````

**What it does:** Joins two strings, first with no space and then with a space added in the middle.


**Why it works:** `+` on two strings *concatenates*: the second string goes on the end of the first. Python never adds a space for you, so you add a `' '` string yourself.


**Output:** `LearningPython` then `Learning Python`

## Repeating strings with `*`

````python
myString = 'Python is great\n'   # \n is a newline character
print(myString * 3)               # * on a string repeats it
````

**What it does:** Prints the line three times, each on its own line, followed by an extra blank line.


**Why it works:** `*` with a string and an int repeats the string that many times. `\n` is a newline, so each copy ends by starting a new line. `print()` also adds its own newline at the end, which is where the extra blank line comes from.


**Output:** Three lines of `Python is great`, then a blank line.


**Notes to self:** `'ha' * 3` → `'hahaha'`. A float doesn't work: `'ha' * 2.5` is a TypeError.

## Escape sequences: `\n`, `\t`, and friends

````python
print("Line one\nLine two")       # \n starts a new line
print("Name:\tEvan")              # \t inserts a tab
print("C:\\Users\\Evan")           # \\ prints one real backslash
print('It\'s done')               # \' puts a ' inside a '...' string
print("She said \"hi\"")          # \" puts a " inside a "..." string
print("It's easier")              # or just use the other kind of quotes
````

**What it does:** Prints text with a line break, a tab, backslashes, and quotes inside quotes.


**Why it works:** A backslash `\` inside a string means "the next character is special." Python swaps the pair for one character: `\n` becomes a line break, `\t` becomes a tab, `\\` becomes a single backslash, and `\'` / `\"` become quote marks that don't end the string.


**Output:**
```
Line one
Line two
Name:	Evan
C:\Users\Evan
It's done
She said "hi"
It's easier
```


**Notes to self:** Same as JavaScript. The `\n` is why the "Repeating strings" example prints each copy on its own line. Easiest fix for quotes: wrap the string in the *other* kind of quote (`"It's"` or `'She said "hi"'`).

## f-strings: putting variables inside text

````python
name = 'Evan'
grade = 85
print(f"{name} got a {grade}")            # f before the quote, variables in { }
print(f"Next year: {grade + 5}")           # any expression works inside { }
print("Old way: " + name + " got a " + str(grade))   # same result, more work
````

**What it does:** Builds sentences that include variable values.


**Why it works:** Putting `f` before the quotes makes it an f-string. Anything inside `{ }` gets evaluated and dropped into the text, and numbers are converted automatically, so no `str()` is needed.


**Output:** `Evan got a 85`, `Next year: 90`, `Old way: Evan got a 85`


**Notes to self:** Same idea as JS template literals: `` `${name} got a ${grade}` ``. Python uses `f"..."` and `{ }` with no `$`. Forgetting the `f` prints the braces literally.

# Part 3: Numbers and types

## Data types and `type()`

````python
print(type(85))        # <class 'int'>
print(type(85.5))      # <class 'float'>
print(type('85'))      # <class 'str'>  quotes make it text, even if it looks like a number
print(type(True))      # <class 'bool'>
````

**What it does:** Shows the type of four different values.


**Why it works:** Every value has a type, and the type decides what operations work on it. `type()` tells you which one you have. `'85'` in quotes is a string, not a number.


**Output:** `<class 'int'>`, `<class 'float'>`, `<class 'str'>`, `<class 'bool'>`


**Notes to self:** When something acts weird, `print(type(x))` is the first thing to check.

---
## Floats (decimal numbers)

````python
price = 19.99                  # a float: has a decimal point
print(type(price))
print(float('3.5') + 1)        # float() converts text (or an int) to a float
print(float(5))                # an int becomes 5.0
print(int(9.99))               # int() on a float CHOPS the decimal, it doesn't round
print(round(9.99))             # round() rounds to the nearest whole number
print(round(3.14159, 2))       # round to 2 decimal places
print(0.1 + 0.2)               # floats aren't perfectly exact
````

**What it does:** Creates a float, converts to and from floats, and rounds.


**Why it works:** A float is any number with a decimal point, even `5.0`. Mixing an int and a float in math gives a float. `float()` converts to a float. `int()` on a float cuts off everything after the decimal, while `round()` actually rounds. Computers store decimals in binary, so some values come out slightly off.


**Output:** `<class 'float'>`, `4.5`, `5.0`, `9`, `10`, `3.14`, `0.30000000000000004`


**Notes to self:** For input with decimals (prices, GPAs), use `float(input(...))` instead of `int(input(...))`. `int('3.5')` crashes with ValueError. `0.1 + 0.2` is the same in JS: it's how computers do decimals, not a Python bug.

## Converting types: `str()` and `int()`

````python
print(str(111) + str(222))     # numbers → strings, so + joins them
print(int('111') + int('222'))  # strings → numbers, so + adds them
print(111 + 222)                # already ints: plain math
````

**What it does:** Shows the same `+` doing joining or adding depending on the types.


**Why it works:** `+` joins strings and adds numbers. `str()` turns a number into text, and `int()` turns numeric text into a whole number. Converting changes what `+` does.


**Output:** `111222`, then `333`, then `333`


**Notes to self:** In my Data-type-conversion notebook, the second example ran `int()` on values that were already ints, so it didn't actually convert anything. The middle line here shows the real string → int conversion.

# Part 4: Input

## Getting input with `input()`

````python
grade = int(input("Enter your grade: "))   # input() returns a str, int() converts it
print(grade + 5)
````

**What it does:** Asks the user for a grade, converts it to a number, and prints the grade plus 5.


**Why it works:** `input()` always returns a string, even if the user types a number. Without `int()`, `grade + 5` would crash because Python won't add a string and an int. Read it inside out: `input()` runs first, then `int()` converts the result.


**Output:** Typing `85` prints `90`.


**Notes to self:** Run input code in a `.py` file (like scratchpad.py), not a notebook. In a notebook it pops up a box at the top of the screen.

# Part 5: Operators

## Arithmetic operators

````python
print(7 + 2)    # add
print(7 - 2)    # subtract
print(7 * 2)    # multiply
print(7 / 2)    # divide: always gives a float
print(7 // 2)   # floor division: divide, then drop the decimal
print(7 % 2)    # modulo: the remainder after dividing
print(7 ** 2)   # power: 7 squared
````

**What it does:** Runs every math operator on 7 and 2.


**Why it works:** `/` always returns a float, even when it divides evenly (`6 / 2` is `3.0`). `//` divides and drops the decimal part, `%` gives what's left over, and `**` raises to a power. Python follows normal math order: `**` first, then `* / // %`, then `+ -`. Use parentheses to change the order.


**Output:** `9`, `5`, `14`, `3.5`, `3`, `1`, `49`


**Notes to self:** Different from JS: JS has no `//` (you'd use `Math.floor(7 / 2)`), and `**` is newer in JS but works the same. `% 2` is the classic even/odd check: `n % 2 == 0` means even.

---
## Assignment operators

````python
score = 10      # = assigns a value
score += 5      # same as: score = score + 5
score -= 3      # same as: score = score - 3
score *= 2      # same as: score = score * 2
print(score)
````

**What it does:** Starts `score` at 10, then updates it three times using shortcut operators.


**Why it works:** `+=`, `-=`, `*=`, `/=` and the rest are shortcuts. They take the current value, do the math, and store the result back in the same variable. `=` is assignment, not equality. Comparing uses `==`.


**Output:** `24` (10 + 5 = 15, 15 − 3 = 12, 12 × 2 = 24)


**Notes to self:** Python has no `++` or `--` like JS. Use `count += 1` instead.

---
## Comparison operators

````python
grade = 85
print(grade > 90)     # greater than
print(grade >= 85)    # greater than or equal
print(grade == 85)    # equal: two = signs
print(grade != 85)    # not equal
````

**What it does:** Compares `grade` four ways. Each comparison produces `True` or `False`.


**Why it works:** Comparisons give back a boolean. That's what `if` checks. `==` compares and `=` assigns, the same as in JS, but Python has no `===`.


**Output:** `False`, `True`, `True`, `False`


**Notes to self:** `>` vs `>=` matters at the edges: with `> 90`, a grade of exactly 90 is not an A.

---
## Logical operators: `and`, `or`, `not`

````python
grade = 85
attended = True
print(grade >= 60 and attended)    # True only if BOTH are True
print(grade >= 90 or attended)     # True if EITHER one is True
print(not attended)                # flips True to False
````

**What it does:** Combines conditions to check more than one thing at once.


**Why it works:** `and` needs both sides to be True. `or` needs at least one. `not` reverses a boolean. They're written as plain English words in Python. These are what you use inside an `if` to check several conditions together, like `if grade >= 60 and attended:`.


**Output:** `True`, `True`, `False`


**Notes to self:** JS uses `&&`, `||`, `!`. Those symbols don't work in Python.

# Part 6: Decisions

## `if` / `elif` / `else` and indentation

````python
grade = 85
if grade >= 90:
    print("You got an A")
elif grade >= 80:              # elif lines up with if
    print("You got a B")
elif grade >= 70:
    print("You got a C")
elif grade >= 60:
    print("You got a D")
else:                          # everything that didn't match above
    print("You got an F")
print("We are done!")          # back at the left edge: runs no matter what
````

**What it does:** Checks the grade from the top down and prints one letter grade, then always prints the last line.


**Why it works:** Python checks each condition in order and runs **only the first block that's True**, then skips the rest. `else` catches anything left over. A line ending in `:` starts a block, and the block is everything indented 4 spaces under it.


**Output:** `You got a B` then `We are done!`


**Notes to self:** `if`, `elif`, and `else` sit at the same level. After a block, press Backspace once to come back out.

---
## Nested `if`

````python
grade = 85
attended = True
if grade >= 60:
    print("Passed")
    if attended:                  # an if inside an if: indented one more level
        print("Perfect attendance bonus")
else:
    print("Failed")
````

**What it does:** Checks the grade first. Only if it passes does it check attendance.


**Why it works:** An `if` can go inside another `if` block. The inner one only runs if the outer one was True. Each level is indented 4 more spaces.


**Output:** `Passed`, `Perfect attendance bonus`


**Notes to self:** Often a nested `if` can be written with `and` instead (`if grade >= 60 and attended:`). Nest when the inner check only makes sense after the outer one passes.

# Part 7: Errors

## Three kinds of errors

This one is concepts, not code to run.

**Syntax error:** the code breaks Python's grammar rules, so nothing runs at all. Examples: a missing `:` after `if`, unclosed quotes or parentheses, wrong indentation. Python points at the line.

**Runtime error (exception):** the syntax is fine, but something fails *while running*. Examples: `int('eighty')` → `ValueError`, `'85' > 90` → `TypeError`, using a variable before creating it → `NameError`, `10 / 0` → `ZeroDivisionError`. The program stops at that line unless it's inside `try`/`except`.

**Logic error:** the code runs with no error message, but gives the wrong answer. Example: using `> 90` when you meant `>= 90`. These are the hardest to find. That's what the debugger is for: step through and check whether each value is what you expected.

**Reading an error:** start at the **last line**. It names the error type and what went wrong. The lines above it show *where* it happened.

---
## Handling bad input with `try` / `except`

````python
try:
    grade = int(input("Enter your grade: "))   # this line can fail, so it goes inside try
    print("Your grade is", grade)
except ValueError:                             # runs only if int() fails
    print("That's not a number.")
print("We are done!")
````

**What it does:** Asks for a grade and prints it. If the user types something that isn't a number, it prints a friendly message instead of crashing.


**Why it works:** `try` runs its block and watches for errors. If one happens, Python jumps straight to the matching `except` block and skips the rest of `try`. `int('eighty')` raises a `ValueError`, so that's the one to catch. Only lines *inside* `try` are protected.


**Output:** `85` → `Your grade is 85`, `We are done!`. `eighty` → `That's not a number.`, `We are done!`


**Notes to self:** Same idea as `try { } catch (err) { }` in JS. `except` needs at least one indented line under it.

### Gotchas

> **Gotcha:** `print` without parentheses prints nothing.
>
> **Why:** `print` alone just refers to the function without calling it, and Python doesn't treat that as an error.
>
> **Fix:** Always call it: `print("text")`.

> **Gotcha:** `elif` or `else` indented under the `if` block.
>
> **Why:** Python reads it as being inside the `if` block, where it doesn't belong.
>
> **Fix:** Line `elif`/`else` up with the `if`. The error is `SyntaxError: invalid syntax`.
