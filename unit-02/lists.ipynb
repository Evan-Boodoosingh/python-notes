## Overview

**Big idea:** A list stores many values in one variable, in order, and you reach each one by its index. It's Python's version of an array, and it's mutable, so you can change it after creating it.

**Key terms:**
- **data collection type:** a group of related values organized so they're easy to work with
- **list:** Python's simplest data collection type, written in square brackets `[ ]`
- **element (item):** one value inside a list
- **index:** an element's position, as an integer. Starts at `0`
- **mutable:** can be changed after it's created (reassign, add, remove elements)
- **nested list:** a list inside another list
- **empty list:** a list with no elements, `[]`
- **mapping:** the relationship between each index and the element it points to

**General notes:**
- Picture a parking lot: the lot is one object, the spots are numbered starting at 0, and each spot holds something (or is empty).
- A strict array (like Java's) needs every element to be the same type. A Python list can mix types, same as a JS array.
- A list of length n has indexes `0` to `n-1`, or `-1` to `-n` counting backward.
- List indexes work exactly like string indexes.

# Part 1: Creating lists

## Making a list

```python
pets = ['dog', 'cat', 'fish']   # strings need quotes
numbers = [17, 123]             # numbers don't
empty = []                      # an empty list
print(pets)
print(numbers)
print(empty)
```

**What it does:** Creates three lists and prints them.


**Why it works:** Square brackets create a list. Elements are separated by commas, and each one is written the same way you'd write it on its own (quotes for strings, none for numbers).


**Output:** `['dog', 'cat', 'fish']`, `[17, 123]`, `[]`


**Notes to self:** Python prints strings in a list with single quotes, even if you typed double quotes.

---
## Mixed types and nested lists

```python
mixed = ['spam', 2.0, 5, [10, 20]]   # str, float, int, and a list inside a list
print(mixed)
print(mixed[3])                       # the nested list is one element
```

**What it does:** Makes a list holding four different types, then prints the nested list on its own.


**Why it works:** Everything in Python is an object, so a list can hold any mix of types. A list inside a list counts as one element.


**Output:** `['spam', 2.0, 5, [10, 20]]`, `[10, 20]`


**Notes to self:** This is the main difference from a strict array, which needs one type for every element.

---
## Commas inside quotes

```python
animals = ['hamster', 'rabbit, dog, cat', 'chipmunk']
print(len(animals))   # len() counts the elements
print(animals[1])
```

**What it does:** Counts the elements and prints the middle one.


**Why it works:** The quotes decide where each element starts and ends, not the commas. `'rabbit, dog, cat'` is one string.


**Output:** `3`, `rabbit, dog, cat`


**Notes to self:** Count elements by matching quote pairs, not by counting commas.

# Part 2: Indexing

## Reading an element by index

```python
pets = ['dog', 'cat', 'fish']
print(pets[0])       # first element is index 0
print(pets[2])       # last of 3 is index 2
print(pets[5 - 4])   # any integer expression works: 5 - 4 is 1
```

**What it does:** Prints three elements using their indexes.


**Why it works:** `listname[x]` gets the element at position `x`. Counting starts at 0, so the last index is always one less than the length.


**Output:** `dog`, `fish`, `cat`


**Notes to self:** Same as JS arrays.

---
## Negative indexes

```python
pets = ['dog', 'cat', 'fish']
print(pets[-1])   # last element
print(pets[-2])
print(pets[-3])   # first element, counted from the back
```

**What it does:** Prints the elements counting backward from the end.


**Why it works:** A negative index counts from the end. `-1` is always the last element, no matter how long the list is.


**Output:** `fish`, `cat`, `dog`


**Notes to self:** Different from JS: `arr[-1]` there gives `undefined`. You'd need `arr.at(-1)`.

---
## Index out of range

```python
pets = ['dog', 'cat', 'fish']
print(pets[3])   # valid indexes are only 0, 1, 2
```

**What it does:** Crashes.


**Why it works:** Asking for an index that doesn't exist raises an `IndexError`. `pets[-4]` fails the same way.


**Output:** `IndexError: list index out of range`


**Notes to self:** Different from JS, where a bad index quietly returns `undefined`. Python stops the program instead.

# Part 3: Changing lists

## Replacing an element

```python
pets = ['dog', 'cat', 'fish']
pets[1] = 'kitten'   # brackets on the LEFT of = pick which element to replace
print(pets)
```

**What it does:** Swaps `'cat'` for `'kitten'`.


**Why it works:** Lists are mutable. When `list[index]` is on the left of `=`, it picks the element to overwrite. On the right, it just reads the value.


**Output:** `['dog', 'kitten', 'fish']`


**Notes to self:** Strings are NOT mutable, so `name[0] = 'X'` would error. Lists are.

### Gotchas

> **Gotcha:** Using `len(list)` as the last index.
>
> **Why:** Indexes start at 0, so a list of 3 ends at index 2. `pets[len(pets)]` is one past the end.
>
> **Fix:** Use `pets[len(pets) - 1]`, or just `pets[-1]`.

### Error log

```python
pets = ['dog', 'cat', 'fish']
print(pets[3])
```

**Error message:** `IndexError: list index out of range`


**What it means:** I asked for a position the list doesn't have.


**Fix:** Stay between `0` and `len(list) - 1`, or `-1` and `-len(list)`.
