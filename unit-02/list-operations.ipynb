## Overview

**Big idea:** Operators, methods, and built-in functions let you check, combine, slice, add, remove, sort, and step through a list's elements. Builds on `lists.ipynb` (Unit 2, Chapter 1).

**Key terms:**
- **slice:** a piece of a list, `list[start:end]`, where `end` is not included
- **method:** a function called on an object with a dot, `pets.append('x')`
- **function:** takes the list as an argument instead, `len(pets)`
- **in place:** changes the original list instead of making a new one
- **iteration:** repeating a process, like going through elements one by one
- **iterable:** an object you can go through one item at a time (lists, strings)
- **iterator:** an object that steps through an iterable and remembers its place

**General notes:**
- Methods like `append`, `insert`, `extend`, `sort`, `remove`, and `pop` change the list in place. `+`, `*`, and slices make new lists.
- Removing: `pop` when you want the value back, `del` by index when you don't, `remove` when you know the value but not the index.
- These operators work the same way on strings: `in`, `+`, `*`, and slicing.

# Part 1: List operators

## Checking membership with `in`

```python
pets = ['dog', 'cat', 'fish']
print('dog' in pets)
print('monkey' in pets)
print('monkey' not in pets)   # the opposite check
```

**What it does:** Checks whether each value is in the list.


**Why it works:** `x in list` returns `True` or `False`. Same operator that checks for text inside a string.


**Output:** `True`, `False`, `True`


**Notes to self:** Like JS `arr.includes('dog')`, but it's an operator instead of a method.

---
## Replacing the whole list by accident

```python
pets = ['dog', 'cat', 'fish']
pets = 'hamster'   # no index, so the whole variable gets replaced
print(pets)
print(type(pets))
```

**What it does:** Overwrites the entire list with one string.


**Why it works:** Without `[index]`, `=` reassigns the variable itself. It's not a list anymore.


**Output:** `hamster`, `<class 'str'>`


**Notes to self:** To change one element you need the index: `pets[1] = 'hamster'`.

---
## Joining lists with `+`

```python
first = [1, 2, 3]
second = [4, 5, 6]
combined = first + second
print(combined)
print(first)   # the originals don't change
```

**What it does:** Makes a new list from two lists.


**Why it works:** `+` on two lists concatenates them, the second one tacked onto the end of the first. It builds a new list and leaves both originals alone.


**Output:** `[1, 2, 3, 4, 5, 6]`, `[1, 2, 3]`


**Notes to self:** In JS, `[1,2] + [3]` turns into the string `'1,23'`. Python actually joins them (like JS `[...a, ...b]`).

---
## Repeating lists with `*`

```python
zeros = [0] * 4
print(zeros)
pets = ['dog', 'cat', 'fish']
print(pets * 2)
```

**What it does:** Repeats each list.


**Why it works:** `list * int` repeats the whole list that many times, same as `*` on a string.


**Output:** `[0, 0, 0, 0]`, `['dog', 'cat', 'fish', 'dog', 'cat', 'fish']`


**Notes to self:** `[0] * 4` is a quick way to make a list of a set size.

---
## Slicing with `[start:end]`

```python
pets = ['dog', 'cat', 'fish', 'rabbit', 'hamster', 'bird']
print(pets[1:3])   # index 1 up to, but NOT including, 3
print(pets[:3])    # no start: from the beginning
print(pets[3:])    # no end: to the end
print(pets[:])     # neither: a copy of the whole list
```

**What it does:** Pulls out pieces of the list.


**Why it works:** `list[a:b]` returns a new list from index `a` up to `b - 1`. The end index is never included. Leave out `a` to start at 0, leave out `b` to go to the end.


**Output:** `['cat', 'fish']`, `['dog', 'cat', 'fish']`, `['rabbit', 'hamster', 'bird']`, `['dog', 'cat', 'fish', 'rabbit', 'hamster', 'bird']`


**Notes to self:** Same as JS `arr.slice(1, 3)`. Trick: `b - a` is how many elements you get.

---
## Replacing several elements with a slice

```python
pets = ['dog', 'cat', 'fish', 'rabbit', 'hamster', 'bird']
pets[1:3] = ['dinosaur', 'robot']   # replaces index 1 and 2
print(pets)
```

**What it does:** Swaps out `'cat'` and `'fish'` in one line.


**Why it works:** A slice on the LEFT of `=` picks a range to overwrite, the same way a single index does. Index 3 (`'rabbit'`) isn't touched because the end is excluded.


**Output:** `['dog', 'dinosaur', 'robot', 'rabbit', 'hamster', 'bird']`

# Part 2: Adding elements

## `.append()`: add one to the end

```python
pets = ['dog', 'cat', 'fish']
pets.append('dinosaur')
print(pets)
```

**What it does:** Adds `'dinosaur'` to the end.


**Why it works:** `.append(x)` changes the list in place by adding one element at the end.


**Output:** `['dog', 'cat', 'fish', 'dinosaur']`


**Notes to self:** JS `push()`.

---
## `.insert()`: add one at a position

```python
pets = ['dog', 'cat', 'fish', 'rabbit']
pets.insert(3, 'dinosaur')   # (index, value)
print(pets)
pets.insert(0, 'bird')       # 0 puts it at the front
print(pets)
```

**What it does:** Adds elements at specific positions.


**Why it works:** `.insert(index, value)` puts the value AT that index and shifts everything after it one spot right. Index comes first, value second.


**Output:** `['dog', 'cat', 'fish', 'dinosaur', 'rabbit']`, `['bird', 'dog', 'cat', 'fish', 'dinosaur', 'rabbit']`


**Notes to self:** The course's output for this example is missing a comma between `'dinosaur'` and `'rabbit'`. That's a typo on their end.

---
## `.extend()`: add every element from another list

```python
first = ['dog', 'cat', 'fish']
second = ['rabbit', 'hamster']
first.extend(second)
print(first)
print(second)   # unchanged
```

**What it does:** Adds both elements of `second` onto the end of `first`.


**Why it works:** `.extend(otherList)` adds each element individually, in place. `second` stays the same.


**Output:** `['dog', 'cat', 'fish', 'rabbit', 'hamster']`, `['rabbit', 'hamster']`


**Notes to self:** Difference from `+`: `extend` changes `first` itself, `+` makes a new list. Difference from `append`: `first.append(second)` would add the whole list as ONE nested element.

# Part 3: Sorting and removing

## `.sort()`: lowest to highest

```python
pets = ['dog', 'cat', 'fish', 'rabbit', 'hamster', 'bird']
pets.sort()
print(pets)
nums = [2, 45, 9, 17, 1, 2]
nums.sort()
print(nums)
```

**What it does:** Sorts strings alphabetically and numbers smallest to largest.


**Why it works:** `.sort()` rearranges the list in place.


**Output:** `['bird', 'cat', 'dog', 'fish', 'hamster', 'rabbit']`, `[1, 2, 2, 9, 17, 45]`


**Notes to self:** Better than JS: JS `sort()` compares numbers as text, so `[2, 45, 9]` becomes `[2, 45, 9]`. Python sorts numbers correctly.

---
## `.pop()`: remove AND get it back

```python
pets = ['dog', 'cat', 'fish', 'rabbit', 'hamster', 'bird']
removed = pets.pop(3)    # remove index 3
print("Remaining:", pets)
print("Removed:", removed)
last = pets.pop()        # no index: removes the last one
print("Removed:", last)
```

**What it does:** Removes elements and saves what was removed.


**Why it works:** `.pop(index)` removes the element at that index and returns it. With no index, it removes the last element.


**Output:** `Remaining: ['dog', 'cat', 'fish', 'hamster', 'bird']`, `Removed: rabbit`, `Removed: bird`


**Notes to self:** Use it when you still need the removed value. No-index version is JS `pop()`.

---
## `del`: remove by index, get nothing back

```python
pets = ['dog', 'cat', 'fish', 'rabbit', 'hamster', 'bird']
del pets[3]
print(pets)
```

**What it does:** Deletes `'rabbit'`.


**Why it works:** `del` is a statement, not a method: no dot, no parentheses. It deletes and returns nothing.


**Output:** `['dog', 'cat', 'fish', 'hamster', 'bird']`


**Notes to self:** The course calls it `.del`, but there's no dot. It's `del pets[3]`.

---
## `.remove()`: remove by value

```python
pets = ['dog', 'cat', 'fish', 'rabbit', 'cat', 'bird']
pets.remove('cat')   # only the FIRST 'cat'
print(pets)
```

**What it does:** Removes the first `'cat'` and leaves the second.


**Why it works:** `.remove(value)` finds the first match and deletes it. If the value isn't there, it raises a `ValueError`.


**Output:** `['dog', 'fish', 'rabbit', 'cat', 'bird']`


**Notes to self:** Pick by what you know: index and want it back = `pop`, index only = `del`, value = `remove`.

## Built-in functions: `len`, `max`, `min`, `sum`

```python
nums = [2, 45, 9, 17, 1, 4]
print("Max:", max(nums))
print("Min:", min(nums))
print("Length:", len(nums))
print("Sum:", sum(nums))
print("Average:", sum(nums) / len(nums))
```

**What it does:** Runs each function on the list and calculates the average.


**Why it works:** Methods are called on the list with a dot (`nums.sort()`). Functions take the list as an argument (`len(nums)`).


**Output:** `Max: 45`, `Min: 1`, `Length: 6`, `Sum: 78`, `Average: 13.0`


**Notes to self:** Average comes out as `13.0` because `/` always returns a float. JS needs `Math.max(...arr)` or `reduce` for these.

# Part 4: Iterables and iterators

## `iter()` and `next()`

```python
nums = [2, 45, 9]
it = iter(nums)    # make an iterator that starts before the first element
print(next(it))
print(next(it))
print(next(it))
```

**What it does:** Steps through the list one element at a time.


**Why it works:** An **iterable** is anything you can go through one item at a time (lists, strings). `iter()` makes an **iterator** from it, which remembers its place. Each `next()` returns the next element and moves forward.


**Output:** `2`, `45`, `9`


**Notes to self:** Roll call: `iter()` picks the line, `next()` asks the next person to sound off. Loops use this behind the scenes, which is coming next chapter.

---
## Running out: `StopIteration`

```python
it = iter('hi')    # strings are iterable too
print(next(it))
print(next(it))
print(next(it))    # nothing left
```

**What it does:** Prints each letter, then crashes.


**Why it works:** Calling `next()` after the last element raises `StopIteration`.


**Output:** `h`, `i`, then `StopIteration`

### Gotchas

> **Gotcha:** Saving the result of `.sort()`.
>
> **Why:** `.sort()` changes the list in place and returns `None`. `nums = nums.sort()` wipes out your list.
>
> **Fix:** Just call `nums.sort()` on its own line.

> **Gotcha:** `append` with a list instead of `extend`.
>
> **Why:** `a.append([4, 5])` adds ONE element, a nested list: `[1, 2, 3, [4, 5]]`.
>
> **Fix:** Use `a.extend([4, 5])` to add each element: `[1, 2, 3, 4, 5]`.

> **Gotcha:** Expecting a slice to include the end index.
>
> **Why:** `pets[1:3]` stops before index 3, so you get 2 elements, not 3.
>
> **Fix:** The end number is one past the last element you want.

> **Gotcha:** `.remove()` on a value that isn't there.
>
> **Why:** It raises `ValueError: list.remove(x): x not in list`.
>
> **Fix:** Check first: `if 'cat' in pets: pets.remove('cat')`.
